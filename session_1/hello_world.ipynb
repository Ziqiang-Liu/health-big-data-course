```python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    "sample_id": [1, 2, 3, 4, 5, 6],
    "group": ["control"]*3 + ["treatment"]*3,
    "activity": [90, 100, 110, 120, np.nan, 140],
})
summary = data.groupby("group")["activity"].agg(["count", "mean"])
summary
```

### 这段代码计算了什么？处理组的平均数使用了几个测量值？

这段代码按 group 分组，计算每组酶活性的有效测量数量（count）和平均值（mean）。
处理组数据为 120、NaN、140。NaN 是缺失值，不参与计算，因此处理组平均数使用了 2 个测量值：

(120 + 140) / 2 = 130

### 如何解决这个问题

如果你指的是处理组有一个 NaN 缺失值，这不是代码错误。
处理方式有三种：
1. 保留缺失值（通常推荐）
   当前代码会忽略 NaN，所以处理组 count=2，平均值为 130。
2. 重新测量缺失样本
   这是最可靠的方法，可以得到真正的 3 个测量值平均数。
3. 进行缺失值插补
   只有在有合理实验依据时使用，不能简单把缺失值填成 0，因为“未测量”不等于“酶活性为 0”。
如果还想同时查看每组总样本数和有效测量数，可以使用：
data.groupby("group")["activity"].agg(["size", "count", "mean"])
其中：
- size：该组总样本数
- count：有效测量值数量
- mean：有效测量值平均数
处理组会显示总数为 3，但有效测量数为 2。
